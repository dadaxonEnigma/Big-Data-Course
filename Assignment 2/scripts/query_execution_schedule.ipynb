```python
import matplotlib.pyplot as plt
import numpy as np
import os

output_dir = 'C:/Users/User/Desktop/2 Part/Big Data/Assignment 2/output/'
os.makedirs(output_dir, exist_ok=True)


queries = ['Q1', 'Q2', 'Q3']
postgresql_avg = [12.8, 0.02, 0.0169]
mongodb_avg = [0.17, 2.8, 0.0035]
neo4j_avg = [0.23, 0.15, 0.02] 

postgresql_std = [0.4, 0.1, 0.005]
mongodb_std = [0.02, 0.2, 0.003]
neo4j_std = [0.03, 0.01, 0.005]

x = np.arange(len(queries))
width = 0.25

fig, ax = plt.subplots(figsize=(10,6))

ax.bar(x - width, postgresql_avg, width, yerr=postgresql_std, capsize=5, label='PostgreSQL')
ax.bar(x, mongodb_avg, width, yerr=mongodb_std, capsize=5, label='MongoDB')
ax.bar(x + width, neo4j_avg, width, yerr=neo4j_std, capsize=5, label='Neo4j')

ax.set_xlabel('Queries')
ax.set_ylabel('Average Execution Time (s)')
ax.set_title('Average Query Execution Times per Database')
ax.set_xticks(x)
ax.set_xticklabels(queries)
ax.legend()

plt.tight_layout()
output_file = os.path.join(output_dir, 'query_execution_times.png')
plt.savefig(output_file)
plt.show()
```
