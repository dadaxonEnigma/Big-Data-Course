```python
from pyspark.sql import SparkSession
import pyspark.sql.functions as F

spark = SparkSession \
    .builder\
    .appName('pyspark_local')\
    .getOrCreate()
print('Активные Spark сессии:', spark.sparkContext.uiWebUrl)
```

## How to create RDDs

```python
df = spark.read.csv('titanic.csv',header=True,inferSchema=True)
```

```python
df.show(2,False,True)
```

```python
result = df\
    .withColumnRenamed('Sex','Gender')\
    .withColumnRenamed('PassengerId','Passenger')
result.printSchema()
```

```python
result.select('Pclass')\
    .distinct()\
    .show()
```

## Group By

```python
result.groupBy('Pclass')\
    .agg(F.count('*')\
    .alias('total_rows'))\
    .orderBy(F.col('total_rows').desc())\
    .show()
```
